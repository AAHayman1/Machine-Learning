## 14. Show boxplots for the numeric features to detect outliers. (4pt)

```python
#TODO

plt.figure(figsize = (45, 15)) 
plt.rc('xtick', labelsize=15)
numdata.boxplot()
plt.show()

normalized_numdata = numdata
for column in numdata.columns :
  normalized_numdata[column] = (numdata[column]-numdata[column].mean())/numdata[column].std()

plt.figure(figsize = (45, 15)) 
plt.rc('xtick', labelsize=15)
normalized_numdata.boxplot()
plt.show()

```
